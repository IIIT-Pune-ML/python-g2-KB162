Find sum of each line, mean, median and standard deviation of all lines in the file.

```
#sum , mean , median ,sd of each line
import statistics

line_sum = []

with open('poem.txt', 'r', encoding='utf-8-sig') as file:
    for i, line in enumerate(file, start=1):
        each_line = line.rstrip('\n')
        length = len(each_line)
        line_sum.append(length)

        print(f"Line {i} length: {length}")

print("mean", statistics.mean(line_sum))
print("median:", statistics.median(line_sum))
print("sd", statistics.stdev(line_sum))
```

# Working with CSV

Let's import a file with fuel economy data

```
import csv
with open('mpg.csv') as csvfile:
  mpg = list(csv.DictReader(csvfile))
print(mpg)


```

Find out how many items are there in the dictionary

```
import csv
with open('mpg.csv') as csvfile:
  mpg = list(csv.DictReader(csvfile))
print(len(mpg))


```

Find the column names with keys

```
import csv
with open('mpg.csv') as csvfile:
  mpg = list(csv.DictReader(csvfile))
print(mpg[0].keys())
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```
total = 0
for car in mpg:
  total += float(car['cty'])
average = total / len(mpg)
print("avg ", average)
```

Find the average hwy fuel economy across all cars

```
average = sum(float(car['hwy']) for car in mpg) / len(mpg)
print("average" ,average)
```

Use set to return the unique values for the number of cylinders the cars.

```
cylinders = set(car['cyl'] for car in mpg)
print(cylinders)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```
groups = {}
for car in mpg:
  cyl = car['cyl']
  cty = float(car['cty'])

  if cyl not in groups :
    groups[cyl] = []
    groups[cyl].append(cty)
for cyl in groups:
  average = sum(groups[cyl]) / len(groups[cyl])
  print(cyl , average)
```

```
MpgByCyl = []

for c in cylinders: # iterate over all the cylinder levels
    summpg = 0
    typecount = 0
    for d in mpg: # iterate over all dictionaries
        if d['cyl'] == c:
          summpg += float(d['cty'])
          typecount += 1
    avg = summpg / typecount

MpgByCyl.append((c, avg))
MpgByCyl.sort(key=lambda x : x[1])
MpgByCyl
```

Return the unique values for the class types in our dataset using set.

```
class_types = set()

for car in mpg:
    class_types.add(car['class'])

print(class_types)
```

Find average miles per gallon in the highway for each class of vehicles

```
MpgByClass = []

for t in class_types:
    sum_mpg = 0
    count = 0

    for d in mpg:
        if d['class'] == t:
            sum_mpg += float(d['hwy'])
            count += 1

    average = sum_mpg / count
    MpgByClass.append((t, average))

print(MpgByClass)
```
